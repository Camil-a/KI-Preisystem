# Librairies


```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

# Load Data
Data ist stored as an Excel file. So, we use Pandas to load data

```python

def load_data(file_path):
    data= pd.read_csv(file_path, sep=";")
    return data

file_path="../Data/useCarData.csv"

data= load_data(file_path)
print(data.head())
```

```python
print(data.columns)
```

First of all, we split test data for later analysis

```python
#x=data.drop(['Price (EUR)','Sale price'], axis=1)
x= data.drop(['Price (EUR)'], axis=1)  # predict Variable

y=data['Price (EUR)']

x_train, x_test, y_train, y_test= train_test_split(x, y, test_size=0.2, random_state=42)
data= x_train.copy()
data['Price (EUR)']= y_train
#print(data.isnull().sum())
```

# Fill missing values with a value
 In the dataset we have 15 missing row Odor, 14 row Daeler rating and 6 with Extras row

```python
data['Odor'] = data['Odor'].fillna("Unknown")
data['Extras'] = data['Extras'].fillna("None")
data['Dealer rating'] = data['Dealer rating'].fillna(data['Dealer rating'].median())

print(data.isnull().sum())
```

## Handle Duplicate Rows

```python
duplicate_rows=data.duplicated()

if duplicate_rows.any():
    print(" There ist Duplicate rows")
    print(f"Number of duplicate rows: {duplicate_rows.sum()}")
    data= data.drop_duplicates()
else:
    print("No duplicate rows")
print(f"Shape of data after removing duplicates: {data.shape}")
```

## Numerical Sanity check

```python
numerical_cols= data.select_dtypes(include=['int64','float64' ]).columns.tolist()
print(data[numerical_cols].columns)
data[numerical_cols].describe()
```
