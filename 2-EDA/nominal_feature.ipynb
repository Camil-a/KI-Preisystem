### ANOVA test

##### How to interpret?

- <span style="color:red"><b>p < 0.05 → significant</b></span>  
  → The ordinal feature affects the price

- <span style="color:red"><b>p ≥ 0.05 → not significant</b></span> 

```python
# ANOVA test for ordinal features
for i in nominal_features :
    formula=f'Q("Price (EUR)") ~ C(Q("{i}"))' # define the formula
    model = ols(formula, data=data).fit() # fit the model
    anova_table = sm.stats.anova_lm(model, typ=2) # ANOVA table
    p_value = anova_table["PR(>F)"][0] # Get p-value
    
    print(f'{i}, p-value: {p_value}')
```
